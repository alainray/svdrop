```python
import os
import torch
import pandas as pd

# Hiperparámetros globales
seeds = [111, 222, 333]
methods = ["erm", "rw_erm", "gdro"]

rows = []

# ============================================================
# 1) CelebA, civilcomments, MultiNLI
# ============================================================
# Para civilcomments y MultiNLI el corr es 0.9 y está codificado en el nombre del método: erm_0.9, etc.
for d in ["CelebA", "civilcomments", "MultiNLI"]:
    for m in methods:
        if d == "CelebA":
            subdir = m          # p.ej. "erm"
            corr = 1.0
        else:
            subdir = f"{m}_0.9" # p.ej. "erm_0.9"
            corr = 0.9

        for s in seeds:
            root = f"{d}/{subdir}/seed_{s}/basis.pt"
            if not os.path.exists(root):
                print(f"[WARN] falta archivo: {root}")
                continue

            a = torch.load(root, map_location="cpu")
            explained = float(a["basis"]["explained_ratio"].sum())

            rows.append({
                "dataset": d,
                "method": m,  # "erm", "rw_erm", "gdro" (sin el _0.9)
                "corr": corr,
                "seed": s,
                "explained_ratio": explained,  # fracción entre 0 y 1
            })

# ============================================================
# 2) CUB
# ============================================================
# Mapea los sufijos 50, 625, 75, 875, 95 a sus corrs originales
CUB_INV = {50: 0.0, 625: 0.25, 75: 0.5, 875: 0.75, 95: 0.9}

for suffix, corr in CUB_INV.items():
    for m in methods:
        subdir = f"{m}_{suffix}"  # p.ej. "erm_50"
        for s in seeds:
            root = f"CUB/{subdir}/seed_{s}/basis.pt"
            if not os.path.exists(root):
                print(f"[WARN] falta archivo: {root}")
                continue

            a = torch.load(root, map_location="cpu")
            explained = float(a["basis"]["explained_ratio"].sum())

            rows.append({
                "dataset": "CUB",
                "method": m,
                "corr": corr,
                "seed": s,
                "explained_ratio": explained,
            })

# ============================================================
# 3) MNISTCIFAR
# ============================================================
mnistcifar_corrs = [0.0, 0.25, 0.5, 0.75, 0.9]

for corr in mnistcifar_corrs:
    for m in methods:
        subdir = f"{m}_{corr}"  # p.ej. "erm_0.0"
        for s in seeds:
            root = f"MNISTCIFAR/{subdir}/seed_{s}/basis.pt"
            if not os.path.exists(root):
                print(f"[WARN] falta archivo: {root}")
                continue

            a = torch.load(root, map_location="cpu")
            explained = float(a["basis"]["explained_ratio"].sum())

            rows.append({
                "dataset": "MNISTCIFAR",
                "method": m,
                "corr": corr,
                "seed": s,
                "explained_ratio": explained,
            })

# ============================================================
# 4) Construir DataFrame y resumir
# ============================================================

df = pd.DataFrame(rows)
df["explained_percent"] = df["explained_ratio"] * 100.0

# --- Tabla: promedio y desviación estándar por dataset + corr + método ---
group_cols = ["dataset", "corr", "method"]  # si NO quieres corr, quítalo de la lista
summary = (
    df.groupby(group_cols)["explained_percent"]
      .agg(mean="mean", std="std", n="count")
      .reset_index()
)

summary
```

```python
import pandas as pd

# 1) A porcentaje
df["explained_percent"] = df["explained_ratio"] * 100.0

# Elige cómo agrupar:
# - Si quieres separar por correlación:
group_cols = ["dataset", "corr", "method"]
# - Si prefieres solo por dataset+method (promediar sobre corr), usa:
# group_cols = ["dataset", "method"]

# 2) Resumen numérico: mean, std, min, max, n
summary = (
    df.groupby(group_cols)["explained_percent"]
      .agg(mean="mean", std="std", min="min", max="max", n="count")
      .reset_index()
)

# 3) Redondear a 2 decimales
for col in ["mean", "std", "min", "max"]:
    summary[col] = summary[col].round(2)

# 4) Columna formateada al estilo de tu ejemplo:
#    80.63\% {\scriptsize (1.19)}
summary["mean_std_cell"] = summary.apply(
    lambda r: f"{r['mean']:.2f}\\% {{\\scriptsize ({r['std']:.2f})}}",
    axis=1
)

# 5) Elegir columnas finales (puedes ajustar el orden)
cols = [c for c in group_cols] + ["mean_std_cell", "min", "max", "n"]
summary_final = summary[cols]

# 6) Exportar a LaTeX SIN escapar backslashes
latex_str = summary_final.to_latex(index=False, escape=False)

with open("pca_explained_summary.tex", "w", encoding="utf-8") as f:
    f.write(latex_str)

print(latex_str)
```
