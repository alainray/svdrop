## Create Table

```python
a = {50: 0.0, 625: 0.25, 75:0.5, 875: 0.75, 95:0.9}
def format_perc(perc):
    return '{:.2f}\%'.format(perc)
def format_float(perc):
    return '{:.2f}'.format(perc)
splits = ['val','test']

'''
- 'means' = overall results per method, get max value from each seed, 
then average results by seed. For results table.
- 'all' = raw results from csvs
- 'seed' = best result for seed
- 'epoch' = results averaged by seed at the epoch level. For plotting.
'''
def create_results_table(dataset, split, metric="worst_acc"):
    results = pd.read_csv(f'{dataset}_FINAL_{split}.csv')
    if dataset != "MultiNLI":
        results['worst_acc'] = results[['avg_acc_group:0','avg_acc_group:1','avg_acc_group:2','avg_acc_group:3']].min(axis=1)
    else:
        results['worst_acc'] = results[['avg_acc_group:0','avg_acc_group:1','avg_acc_group:2','avg_acc_group:3','avg_acc_group:4','avg_acc_group:5']].min(axis=1)
    epoch_df = results.groupby(['method','corr',"epoch"]).agg(
        {'worst_acc': 'mean','avg_acc': 'mean'}
    ).reset_index()
    grouped_df = results.groupby(['method','corr','seed']).agg(
        {'avg_acc': 'max', 'worst_acc': 'max'}
    ).reset_index()
    means_df = grouped_df.groupby(['method','corr']).agg(
        {'avg_acc':'mean','worst_acc':'mean'}
    ).reset_index()
    for m in ['worst_acc','avg_acc']:
        means_df[m] = 100*means_df[m]
        means_df[f'{m}_std'] = 100*grouped_df.groupby(['method','corr'])[m].std().reset_index()[m]
        means_df['count'] = grouped_df.groupby(['method','corr'])[m].count().reset_index()[m]
        means_df[m] = means_df[m].apply(format_perc)
        means_df[f'{m}_std'] = means_df[f'{m}_std'].apply(format_float)
        means_df[f'final_{m}'] = (means_df[m]).astype(str) + " $\pm$ " + means_df[f'{m}_std'].astype(str) + " (" + means_df['count'].astype(str) + ")"
    return {'means': means_df, 'seed': grouped_df, 'all': results, 'epoch': epoch_df}

def create_final_table(df, metric="avg_acc"):
    final_table = df.pivot(index='method',
                           columns='corr',
                           values=([f"final_{metric}"])).reset_index()

    return final_table
```

```python
dataset = "MultiNLI"
results = dict()
splits = ['val','test']
for s in splits:
    results[s] =create_results_table(dataset, s)
    display(create_final_table(results[s]['means'],"worst_acc"))
    display(create_final_table(results[s]['means'],"avg_acc"))
```

```python
import matplotlib.pyplot as plt
corrs  = [0.0, 0.25, 0.5, 0.75, 0.9, 1.0]

data = [v['epoch'] for v in results.values()]
#data = [train_rs_v[-1],val_rs_v[-1],test_rs_v[-1]]
def plot_methods(ax, df, corr, methods = ["erm"], split="train", metric="avg_acc"):
    data = df[df['corr'] == corr]
  #  fig, ax = plt.subplots(nrows=1,ncols=1)
    for m in methods:
        d = data[data['method']==m]
        x = d['epoch']
        y = d[metric]
        ax.plot(x,y,label=m)
    ax.legend()    
    ax.set_title(f"{split.capitalize()} Accuracy vs Epochs - Corr: {corr}")
    return ax

fig, axs = plt.subplots(nrows=len(corrs), ncols=len(splits), figsize=(len(corrs)*4,30),sharey="row")
for i, corr in enumerate(corrs):
    for j, split in enumerate(splits):
        axs[i,j] = plot_methods(axs[i,j],
                          data[j],
                          corr,
                          methods=["erm","gdro","reweight"],
                          split=split,
                        metric = "avg_acc")
fig.show()
```

```python

def update_state_dict(old, new, layers=["layer4"]):
    # old and new must have same keys
    keys1 = set(old.keys())
    keys2 = set(old.keys())
    assert keys1 == keys2, "Dictionaries do not have the same keys"
    results = {k:v for k,v in old.items()}
    for k, v in new.items():
        for layer in layers:
            if layer in k:
                results[k] = v # update with new value
    return results

r = update_state_dict(m_sd, m1_sd, layers=['layer1',"layer2"])
print(m_sd['layer1.0.conv3.weight'][0][0],
m1_sd['layer1.0.conv3.weight'][0][0],
     r['layer1.0.conv3.weight'][0][0],
     m_sd['layer2.0.conv3.weight'][0][0],
m1_sd['layer2.0.conv3.weight'][0][0],
     r['layer2.0.conv3.weight'][0][0])
```
